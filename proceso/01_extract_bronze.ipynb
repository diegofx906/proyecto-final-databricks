```python
# Databricks notebook source
# MAGIC %md
# MAGIC ### Capa Bronze: Extracción de datos crudos
# MAGIC Leemos los archivos CSV desde el Volumen y los escribimos como tablas Delta.

from pyspark.sql.functions import current_timestamp, lit

# 1. Definir rutas de los archivos subidos
ruta_ventas = "/Volumes/ecommerce_catalog/bronze/raw_data/Ecommerce_Sales_Prediction_Dataset.csv"
ruta_pedidos = "/Volumes/ecommerce_catalog/bronze/raw_data/orders.csv"

# 2. Leer los archivos CSV con encabezado e inferencia de esquema
df_ventas = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_ventas)
df_pedidos = spark.read.option("header", "true").option("inferSchema", "true").option("sep", ";").csv(ruta_pedidos)

# 3. Añadir una columna de auditoría (fecha de ingesta) y escribir en tablas Delta
df_ventas.withColumn("ingestion_date", current_timestamp()).write.mode("overwrite").saveAsTable("ecommerce_catalog.bronze.ventas_raw")
df_pedidos.withColumn("ingestion_date", current_timestamp()).write.mode("overwrite").saveAsTable("ecommerce_catalog.bronze.pedidos_raw")

print("Capa Bronze completada. Tablas creadas: ventas_raw, pedidos_raw")
```
