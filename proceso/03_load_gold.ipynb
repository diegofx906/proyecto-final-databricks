```python
# Databricks notebook source
# MAGIC %md
# MAGIC ### Capa Gold: Agregaciones de Negocio

from pyspark.sql.functions import sum, avg, count, col

# 1. Leer tablas Silver
df_ventas = spark.table("ecommerce_catalog.silver.ventas_clean")
df_pedidos = spark.table("ecommerce_catalog.silver.pedidos_clean")

# 2. Agregación 1: Ventas totales y ticket promedio por categoría de producto
df_gold_ventas_categoria = df_ventas.groupBy("categoria_producto").agg(
    sum("precio").alias("ingreso_total"),
    avg("precio").alias("precio_promedio"),
    sum("unidades_vendidas").alias("unidades_totales"),
    avg("descuento").alias("descuento_promedio")
)

# 3. Agregación 2: Comportamiento de pedidos por hora del día
df_gold_pedidos_hora = df_pedidos.groupBy("hora_pedido").agg(
    count("id_pedido").alias("cantidad_pedidos"),
    avg("order_number").alias("numero_pedido_promedio")
).orderBy("hora_pedido")

# 4. Escribir tablas Gold
df_gold_ventas_categoria.write.mode("overwrite").saveAsTable("ecommerce_catalog.gold.ventas_por_categoria")
df_gold_pedidos_hora.write.mode("overwrite").saveAsTable("ecommerce_catalog.gold.pedidos_por_hora")

print("Capa Gold completada. Tablas creadas: ventas_por_categoria, pedidos_por_hora")
```
